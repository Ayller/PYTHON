```python

# No terminal, uma vez: python -m pip install xlrd pandas
import pandas as pd
```

```python
ARQUIVO = "Tabela 1.1.1.xls"

print("Arquivo:", ARQUIVO)
```

```python
bruto = pd.read_excel(
    ARQUIVO,
    sheet_name="2016",
    header=None,
    engine="xlrd"
)

display(bruto.head(12))
```

```python

dados = bruto.iloc[8:41].copy()

dados.columns = [
    "uf_regiao",
    "total",
    "total_branca",
    "total_preta_parda",
    "homem_branca",
    "homem_preta_parda",
    "mulher_branca",
    "mulher_preta_parda",
]

dados = dados.reset_index(drop=True)

for coluna in dados.columns[1:]:
    dados[coluna] = pd.to_numeric(dados[coluna], errors="coerce")

display(dados)
```

```python
# A média nacional é o valor da própria linha Brasil.
media_nacional = dados.loc[
    dados["uf_regiao"] == "Brasil",
    "total"
].item()

print(f"Média nacional (Brasil): {media_nacional:.2f} horas/semana")
```

### 1

**Qual(is) estado(s) está(ão) acima da média nacional de afazeres domésticos?**

```python
regioes = ["Norte", "Nordeste", "Sudeste", "Sul", "Centro-Oeste"]

estados = dados[
    ~dados["uf_regiao"].isin(["Brasil"] + regioes)
].copy()

acima_media = estados.loc[
    estados["total"] > media_nacional,
    ["uf_regiao", "total"]
].sort_values("total", ascending=False)

print(f"Média nacional: {media_nacional:.2f} horas/semana")
print(f"Estados acima da média: {len(acima_media)}")
                
display(acima_media)
```

### 2

**Quais estados estão abaixo da média?**

```python
abaixo_media = estados.loc[
    estados["total"] < media_nacional,
    ["uf_regiao", "total"]
].sort_values("total", ascending=False)

print(f"Média nacional: {media_nacional:.2f} horas/semana")
print(f"Estados abaixo da média: {len(abaixo_media)}")

display(abaixo_media)
```

## PARTE 4 — Comparação entre homens e mulheres

### 3

**Qual é o percentual de diferença no Brasil entre o tempo médio de mulheres e homens dedicado aos cuidados de pessoas e/ou afazeres domésticos?**

```python
brasil = dados.loc[dados["uf_regiao"] == "Brasil"].iloc[0]

# Calculamos a média dos dois grupos de cor/raça disponíveis para homens e mulheres.
homem = (brasil["homem_branca"] + brasil["homem_preta_parda"]) / 2
mulher = (brasil["mulher_branca"] + brasil["mulher_preta_parda"]) / 2

percentual = (mulher - homem) / homem * 100

print(f"Homens (média das cores):   {homem:.2f} h/semana")
print(f"Mulheres (média das cores): {mulher:.2f} h/semana")
print(f"Diferença: {mulher - homem:.2f} h/semana")
print(f"Percentual de diferença: {percentual:.1f}%")

perc_branca = (brasil["mulher_branca"] / brasil["homem_branca"] - 1) * 100
perc_preta_parda = (brasil["mulher_preta_parda"] / brasil["homem_preta_parda"] - 1) * 100

print(f"Branca: mulheres +{perc_branca:.1f}% em relação aos homens brancos")
print(f"Preta ou parda: mulheres +{perc_preta_parda:.1f}% em relação aos homens pretos ou pardos")
```

## PARTE 5 — Perguntas finais

### 4

**Quantas horas mulheres pretas ou pardas trabalham a mais que mulheres brancas em Minas Gerais?**

```python
minas_gerais = dados.loc[
    dados["uf_regiao"] == "Minas Gerais"
].iloc[0]

diferenca_mg = (
    minas_gerais["mulher_preta_parda"]
    - minas_gerais["mulher_branca"]
)

print(f"Mulheres brancas: {minas_gerais['mulher_branca']:.2f} h/semana")
print(f"Mulheres pretas ou pardas: {minas_gerais['mulher_preta_parda']:.2f} h/semana")
print(f"Diferença: {diferenca_mg:.2f} horas/semana")
```

### 5

**Quantas horas mulheres pretas ou pardas trabalham a mais que homens brancos no Brasil?**

```python
diferenca_brasil = (
    brasil["mulher_preta_parda"]
    - brasil["homem_branca"]
)

print(f"Mulheres pretas ou pardas: {brasil['mulher_preta_parda']:.2f} h/semana")
print(f"Homens brancos: {brasil['homem_branca']:.2f} h/semana")
print(f"Diferença: {diferenca_brasil:.2f} horas/semana")
```

```python
amapa = dados.loc[
    dados["uf_regiao"] == "Amapá"
].iloc[0]

diferenca_amapa = (
    amapa["mulher_preta_parda"]
    - amapa["mulher_branca"]
)

print(f"Mulheres brancas: {amapa['mulher_branca']:.2f} h/semana")
print(f"Mulheres pretas ou pardas: {amapa['mulher_preta_parda']:.2f} h/semana")
print(f"Diferença: {diferenca_amapa:.2f} horas/semana")
```
